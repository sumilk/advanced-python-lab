# Python: Intermediate → Advanced

A hands-on course. Every lesson is a notebook: read, run the cells, change them, break them, then solve the exercises in place.

**How to use a lesson notebook**

1. Run the setup cell at the top (it wires up `check`).
2. Run each demo cell and *read the output* — then edit it and re-run.
3. Exercise cells are yours to fill in; the cell right after grades it.
4. Hints and reference solutions are in collapsed `<details>` blocks.

### Track your progress

Progress is shared across the notebook and terminal front-ends — solving an exercise in either records it.

````python
import sys, pathlib
_h = pathlib.Path.cwd()
sys.path.insert(0, str(next(p for p in (_h, *_h.parents)
                            if (p / 'tutorial').is_dir())))
from tutorial.progress import Progress
from tutorial.registry import all_lessons

p = Progress.load()
for les in all_lessons():
    done = p.exercise_count(les.key)
    total = len(les.exercises())
    bar = '█' * done + '·' * (total - done)
    print(f'{les.number:02d} {les.title:<42} {bar} {done}/{total}')
````
